```python
def sat_to_2sat(n, clauses):
    fresh = n + 1
    nclauses = []
    for clause in clauses:
        if len(clause) == 2:
            nclauses.append(clause)
            continue
        v = fresh
        fresh += 1
        for lit in clause:
            nclauses.append((-lit, -v))
        for _ in range(n):
            nclauses.append((v, -fresh))
            fresh += 1
    return fresh - 1, nclauses



# -------------------------
# SAT HARDCODED
# -------------------------


n = 3  # numero de variables


clauses = [
    [1, -2, 3]
]


new_n, new_clauses = sat_to_2sat(n, clauses)


print("Variables totales:", new_n)
print("Clausulas 2-SAT:")


for i, c in enumerate(new_clauses):
    if i == len(new_clauses) - 1:
        print("{" + ", ".join(map(str, c)) + "}")
    else:
        print("{" + ", ".join(map(str, c)) + "}" + ",")
```

```python
import re


# Lista para almacenar los datos de las variables
all_variables = []

print("Introduce todos los valores de las variables en una sola línea (ej. Variable 1: TRUE, Variable 2: FALSE, ... o Variable 1: TRUE  Variable 2: TRUE ...):")
full_input_string = input("Entrada de variables: ").strip()

# Usar una expresión regular para encontrar pares 'Variable N: VALUE'
# Esto permite manejar diferentes separadores como comas, espacios o saltos de línea entre pares.
# El patrón busca 'Variable' seguido de uno o más dígitos, luego un ':' y 'TRUE' o 'FALSE'.
variable_pattern = re.compile(r'(Variable\s*\d+):\s*(TRUE|FALSE)', re.IGNORECASE)
matches = variable_pattern.finditer(full_input_string)

for match in matches:
    var_name = match.group(1).strip() # Capture 'Variable N'
    raw_value = match.group(2).strip().upper() # Capture 'TRUE' or 'FALSE'

    if raw_value in ['TRUE', 'FALSE']:
        bool_value = True if raw_value == 'TRUE' else False
        all_variables.append({'name': var_name, 'value': bool_value})
    else:
        # Esto no debería ocurrir si el regex es correcto, pero se mantiene como precaución
        print(f"Advertencia: Valor no válido para '{var_name}'. Ignorando.")

if not all_variables:
    print("No se detectaron variables válidas en la entrada. Asegúrate de usar el formato 'Variable N: TRUE/FALSE'.")

print("\n--- Variables seleccionadas (primeras n) ---")
# Seleccionar y mostrar las primeras 'n' variables
for j in range(min(n, len(all_variables))):
    var = all_variables[j]
    # Convertir el valor booleano de nuevo a la cadena 'TRUE' o 'FALSE' para la salida
    output_value = 'TRUE' if var['value'] else 'FALSE'
    print(f"{var['name']}: {output_value}")
```
