```python
def sat_to_2sat(n, clauses):
    fresh = n + 1
    nclauses = []
    for clause in clauses:
        if len(clause) == 2:
            nclauses.append(clause)
            continue
        v = fresh
        fresh += 1
        for lit in clause:
            nclauses.append((-lit, -v))
        for _ in range(n):
            nclauses.append((v, -fresh))
            fresh += 1
    return fresh - 1, nclauses



# -------------------------
# SAT HARDCODED
# -------------------------


n = 3  # numero de variables


clauses = [
    [1, -2, 3]
]


new_n, new_clauses = sat_to_2sat(n, clauses)


print("Variables totales:", new_n)
print("Clausulas 2-SAT:")


for i, c in enumerate(new_clauses):
    if i == len(new_clauses) - 1:
        print("{" + ", ".join(map(str, c)) + "}")
    else:
        print("{" + ", ".join(map(str, c)) + "}" + ",")
```
